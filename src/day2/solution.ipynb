## Example

```python
# example = '1698522-1698528'
example = '11-22,95-115,998-1012,1188511880-1188511890,222220-222224,1698522-1698528,446443-446449,38593856-38593862,565653-565659,824824821-824824827,2121212118-2121212124'

id_rows = [id.split('-') for id in example.split(',')]

def isInvalid(num):
  num_str = str(num)
  str_len = len(num_str)
  first_half = num_str[0:str_len//2]
  second_half = num_str[str_len//2:]
  return first_half == second_half

def find_invalids(ranges):
  invalid_ids = []
  for lower, upper in ranges:
    num_range = range(int(lower), int(upper) + 1)
    for num in num_range:
      if isInvalid(num): invalid_ids.append(num)

  return sum(invalid_ids)
    

find_invalids(id_rows)
```

## Part 1

```python
input_txt = open('./input.txt', 'r').read()
input = [id.split('-') for id in input_txt.split(',')]
find_invalids(input)
```

# Part 2

```python
def is_invalid_deluxe(num):
  num_str = str(num)
  sub_str = ''
  # if you iterate half of the string and find nothing, its not invalid
  # because there is no way at this point there could be a repeating sub str
  for i in range((len(num_str)//2)):
    sub_str += num_str[i]
    str_len = len(num_str.replace(sub_str, ""))
    if (str_len == 0): return True
  
  return False

def find_invalids_delux(ranges):
  invalid_ids = []
  for lower, upper in ranges:
    num_range = range(int(lower), int(upper) + 1)
    for num in num_range:
      if is_invalid_deluxe(num): invalid_ids.append(num)

  return sum(invalid_ids)

find_invalids_delux(input)
```
