## Example

```python

example = ['L68', 'L30', 'R48', 'L5', 'R60', 'L55', 'L1', 'L99', 'R14', 'L82']
example
```

```python
# 0-99 dial that starts at 50
def spin_dial(directions):
  zero_count = 0
  dial = range(100)
  dial_position = 50
  for turn in directions:
    val = int(turn[1:])
    dir = 1 if turn[0] == 'R' else -1
    delta = (val * dir) + dial_position
    dial_position = delta%len(dial)
    if (dial_position == 0):
      zero_count += 1


  return zero_count

spin_dial(example)
```

## Part 1

```python
input = open('./input.txt', 'r').read().split('\n')
spin_dial(input)
```

# Part 2

```python
def spin_dial_delux(directions):
  dial_len = 100
  dial = range(dial_len)
  starting_position = 50
  zero_count = 0
  for turn in directions:
    dir = 1 if turn[0] == 'R' else -1
    delta_abs = int(turn[1:])
    delta = delta_abs * dir
    next_position = (starting_position + delta)%len(dial)
    if (starting_position != 0):
      # if it did not start at zero but ended at 0
      if (next_position != starting_position and next_position == 0):
        zero_count += 1
      # if moved right and passed 0
      elif (dir > 0 and next_position < starting_position):
        zero_count +=1
      # if moved left and passed 0
      elif (dir < 0 and next_position > starting_position):
        zero_count +=1
    # number of times spun
    if (delta_abs > dial_len):
      zero_count += delta_abs // dial_len

    starting_position = next_position
  return zero_count

  

spin_dial_delux(input)
```
