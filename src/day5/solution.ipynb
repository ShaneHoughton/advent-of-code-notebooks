## Example

```python
example = '3-5 10-14 16-20 12-18\n1 5 8 11 17 32'
example_ranges, example_ids = [item.split(' ') for item in  example.split('\n')]
```

```python
def is_between_inclusive(num, range):
  lower, upper = [int(bound) for bound in range.split('-')]
  return upper >= num >= lower

def is_fresh(id, ranges):
  for r in ranges:
    if is_between_inclusive(id, r):
      return True
  return False

def get_total_fresh(ranges, ids):
  return len([ id for id in ids if is_fresh(int(id), ranges) ])

get_total_fresh(example_ranges, example_ids)
```

# Part 2

```python
# helpers
def has_overlap(range1, range2):
  lower1, upper1 = range1
  lower2, upper2 = range2
  return upper1 >= lower2 and lower1 <= upper2

def merge_ranges(range1, range2):
  lower1, upper1 = range1
  lower2, upper2 = range2
  return (min(lower1, lower2), max(upper1, upper2))

def parse_range(range_str):
  bound1, bound2 = range_str.split('-')
  return int(bound1), int(bound2)

def parse_and_sort_ranges(range_str_list):
  parsed_ranges = [ parse_range(r) for r in range_str_list]
  parsed_ranges.sort()
  return parsed_ranges

def get_range_size(range):
  lower, upper = range
  return upper - lower + 1

def reduce_ranges(final_ranges, remaining_ranges):
  if len(remaining_ranges) == 1:
    final_ranges.extend(remaining_ranges)
    return final_ranges
  
  remaining_ranges_p = []
  current_range, next_range, *rest = remaining_ranges
  if (has_overlap(current_range, next_range)):
    merged_range = merge_ranges(current_range, next_range)
    remaining_ranges_p.append(merged_range)
    remaining_ranges_p.extend(rest)
  else:
    final_ranges.append(current_range)
    remaining_ranges_p.append(next_range)
    remaining_ranges_p.extend(rest)
  # recursion !!!
  return reduce_ranges(final_ranges, remaining_ranges_p)
```

```python
def get_total_fresh_deluxe(ranges):
  # sorting makes it easier to merge ranges
  sorted_ranges = parse_and_sort_ranges(ranges)
  final_ranges = reduce_ranges([], sorted_ranges)
  return sum([ get_range_size(r) for r in final_ranges ])

get_total_fresh_deluxe(example_ranges)
```
