## Example

```python
example = '3-5 10-14 16-20 12-18\n1 5 8 11 17 32'
example_ranges, example_ids = [item.split(' ') for item in  example.split('\n')]
```

```python
def is_between_inclusive(num, range):
  lower, upper = [int(bound) for bound in range.split('-')]
  return upper >= num >= lower

def is_fresh(id, ranges):
  for r in ranges:
    if is_between_inclusive(id, r):
      return True
  return False

def get_total_fresh(ranges, ids):
  return len([ id for id in ids if is_fresh(int(id), ranges) ])

get_total_fresh(example_ranges, example_ids)
```

```python
input_txt = open('./input.txt', 'r').read()
p1_ranges, p2_ids = [item.split('\n') for item in  input_txt.split('\n\n')]

get_total_fresh(p1_ranges, p2_ids)
```
