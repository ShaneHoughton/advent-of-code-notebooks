## Example

```python
example_txt = open('./example.txt', 'r').read()
def get_points(input):
  return [ tuple([int(coord) for coord in num.split(',')]) for num in input.split()]
example_points = get_points(example_txt)
example_points
```

## Part 1

```python
def calculate_distance(p1, p2):
  def diff_squared(n,m):
    return (n - m) ** 2
  x1, y1, z1 = p1
  x2, y2, z2 = p2
  return (diff_squared(x2, x1) + diff_squared(y2, y1) + diff_squared(z2, z1)) ** 1/2

def calculate_distances(points):
  distances = {}
  for i in range(len(points)):
    coord_i = points[i]
    for j in range(i + 1, len(points)):
      coord_j = points[j]
      calculated_distance = calculate_distance(coord_i,coord_j)
      coord_set = { coord_i, coord_j }
      # in case there are duplicate distances
      if (calculated_distance not in distances): 
        distances[calculated_distance] = [coord_set]
      else:
        distances[calculated_distance].append(coord_set)
  return distances


def create_circuits(distances_dict, num_connections):
  circuits = []
  def find_and_extract_circuit(point):
    for i in range(len(circuits)):
      current_circuit = circuits[i]
      if point in current_circuit: return circuits.pop(i)
    return { point }
  
  distance_keys = list(distances_dict.keys())
  distance_keys.sort()
  for key in distance_keys[0:num_connections]:
    for point_set in distances_dict[key]:
      p1, p2 = point_set
      found_circuit_a = find_and_extract_circuit(p1)
      found_circuit_b = find_and_extract_circuit(p2)
      unioned = found_circuit_a | found_circuit_b
      circuits.append(unioned)
  return circuits

def factor(num_arr):
  value = 1
  for num in num_arr:
    value *= num
  return value

def get_circuit_values(points, num_connections):
  distances = calculate_distances(points)
  circuits = create_circuits(distances, num_connections)
  circuit_lengths = [ len(circuit) for circuit in circuits ]
  circuit_lengths.sort()
  return factor(circuit_lengths[-3:])

get_circuit_values(example_points, 10)
```

```python
def find_last_two_circuits(distances_dict, num_connections_needed):
  circuits = []
  def find_and_extract_circuit(point):
    for i in range(len(circuits)):
      current_circuit = circuits[i]
      if point in current_circuit: return circuits.pop(i)
    return { point }
  
  distance_keys = list(distances_dict.keys())
  distance_keys.sort()
  for key in distance_keys:
    for point_set in distances_dict[key]:
      p1, p2 = point_set
      found_circuit_a = find_and_extract_circuit(p1)
      found_circuit_b = find_and_extract_circuit(p2)
      unioned = found_circuit_a | found_circuit_b
      circuits.append(unioned)
      if (len(circuits) == 1 and len(circuits[0]) == num_connections_needed):
        return p1, p2

def get_circuit_values_deluxe(points):
  distances = calculate_distances(points)
  p1, p2 = find_last_two_circuits(distances, len(points))
  x1, *_ = p1
  x2, *_ = p2
  return x1 * x2

get_circuit_values_deluxe(example_points)
```
