## Example

```python
example = [
  '123 328  51 64 ',
  ' 45 64  387 23 ',
  '  6 98  215 314',
  '*   +   *   +  +'
]

operator_dict = {
  '*': {'fn': lambda a,b: a*b, 'base': 1 },
  '+': {'fn': lambda a,b: a+b, 'base': 0 }
}
def get_operator_info(key):
  value = operator_dict[key]
  return value['fn'], value['base']

def create_matrix(input):
  matrix = [[ int(num) for num in line.split()] for line in input[:-1]]
  matrix.extend([input[-1].split()])
  return matrix

def transpose_matrix(matrix): # could use numpy but wanted to do it myself
  num_rows = len(matrix)
  num_cols = len(matrix[0])
  matrix_p = [[0] * num_rows for _ in range(num_cols)]
  for i in range(num_rows):
    for j in range(num_cols):
      matrix_p[j][i] = matrix[i][j]
  return matrix_p

def do_operations(matrix):
  final_values = []
  for row in matrix:
    fn, final_value = get_operator_info(row[-1])
    for value in row[:-1]:
      final_value = fn(final_value, value)
    final_values.append(final_value)
  return sum(final_values)

example_matrix = create_matrix(example)
transposed_matrix = transpose_matrix(example_matrix)
do_operations(transposed_matrix)
```

## Part 1

```python
input_txt = open('./input.txt', 'r').read()
input = input_txt.split('\n')
p1_matrix = transpose_matrix(create_matrix(input))
do_operations(p1_matrix)
```
