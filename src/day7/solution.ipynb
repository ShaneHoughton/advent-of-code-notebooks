## Example

```python
example_txt = open('./example.txt', 'r').read()
def create_grid(input):
  return [list(coord) for coord in input.split()]
```

```python
class Beam:
  node_paths = {}
  split_count = 0
  total_possible_paths = 0
  def __init__(self, grid):
    self.grid = grid
    self.start_pos = self.find_start_pos()
  
  def find_start_pos(self):
    grid = self.grid
    for i in range(len(grid) - 1):
       for j in range(len(grid[i]) - 1):
        if grid[i][j] == 'S': return i,j

  def start(self):
    self.total_possible_paths = self.traverse_down(self.start_pos)

  def traverse_down(self, current_node):
    # knocks p1 and 2 out by counting total paths and counting each split
    node_paths = self.node_paths
    grid = self.grid
    y,x = current_node
    current_val = grid[y][x]
    at_bottom = y == len(grid) - 1
    if (at_bottom): return 1
    if (current_node in node_paths):
      return node_paths[current_node]
    if (current_val == '.'):
      return self.traverse_down((y + 1, x))
    x_left  = x - 1
    x_right = x + 1
    self.split_count += 1
    node_paths[current_node] = self.traverse_down((y, x_left)) + self.traverse_down((y, x_right))
    return node_paths[current_node]
    
    
my_grid = create_grid(example_txt)
test = Beam(my_grid)
value = test.start()
test.split_count
```
