## Part 1

```python
import copy

def is_accessible(coordinate, matrix):
  y,x = coordinate
  if (matrix[y][x] != '@'): return False

  block_count = 0
  at_top = y == 0
  at_bottom = y == len(matrix) - 1
  at_left = x == 0
  at_right = x == len(matrix[0]) - 1

  left =         '.' if at_left else matrix[y][x - 1]
  top_left =     '.' if at_top or at_left else matrix[y - 1][x - 1]
  top =          '.' if at_top else matrix[y - 1][x]
  top_right =    '.' if at_top or at_right else matrix[y - 1][x + 1]
  right =        '.' if at_right else matrix[y][x + 1]
  bottom_right = '.' if at_bottom or at_right else  matrix[y + 1][x + 1]
  bottom =       '.' if at_bottom else matrix[y + 1][x]
  bottom_left =  '.' if at_bottom or at_left else matrix[y + 1][x - 1]
  
  if (top == '@'): block_count += 1
  if (top_right == '@'): block_count += 1
  if (right == '@'): block_count += 1
  if (bottom_right == '@'): block_count += 1
  if (bottom == '@'): block_count += 1
  if (bottom_left == '@'): block_count += 1
  if (left == '@'): block_count += 1
  if (top_left == '@'): block_count += 1

  return block_count < 4
 

def find_removable_rolls(_matrix):
  matrix = copy.deepcopy(_matrix)
  marked_matrix = copy.deepcopy(_matrix)
  row_len = len(matrix)
  col_len = len(matrix[0])
  rolls_removed = 0
  for y in range(row_len):
    for x in range(col_len):
      if (is_accessible((y,x), matrix)):
        rolls_removed += 1
        marked_matrix[y][x] = 'x'
  marked_matrix = [ "".join(row) for row in marked_matrix]
  return rolls_removed, marked_matrix

exaple_rolls_removed, example_marked_matrix = find_removable_rolls(example)
exaple_rolls_removed
```

# Part 2

```python
def find_removable_rolls_deluxe(_matrix):
  matrix = copy.deepcopy(_matrix)
  marked_matrix = copy.deepcopy(_matrix)
  row_len = len(matrix)
  col_len = len(matrix[0])
  total_removable_roll_count = 0
  current_removable_roll_count = 0
  matrix_iterations = []
  while True:
    for y in range(row_len):
      for x in range(col_len):
        if (is_accessible((y,x), matrix)):
          current_removable_roll_count += 1
          marked_matrix[y][x] = 'x'
    matrix_iterations.append([ "".join(row) for row in marked_matrix])
    matrix = marked_matrix
    marked_matrix = copy.deepcopy(matrix)
    if (current_removable_roll_count > 0):
      total_removable_roll_count += current_removable_roll_count
      current_removable_roll_count = 0
    else:
      break
  
  return total_removable_roll_count, matrix_iterations

exaple_num_spots, example_iterations = find_removable_rolls_deluxe(example)
exaple_num_spots
```
