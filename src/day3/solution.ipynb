## Example

```python
example = ['987654321111111','811111111111119','234234234234278','818181911112111']
banks = [list(b) for b in example]

def get_highest_jolt(bank):
  bank_list = list(bank)
  bank_size = len(bank)
  largest_jolt = 0
  for i in range(0, bank_size):
    for j in range(bank_size - 1, i, -1):
      current_jolt = int(bank_list[i] + bank_list[j])
      largest_jolt = current_jolt if current_jolt > largest_jolt else largest_jolt
  return largest_jolt

def find_batteries(banks):
  jolts = [get_highest_jolt(bank) for bank in banks]
  return sum(jolts)
    
find_batteries(banks)
```

# Part 2

```python
def find_highest_jolt_deluxe(_value_str, target_length, quiet):
  value_str = _value_str
  built_str = ''
  while True:
    current_value_list = [int(val) for val in list(value_str)]
    current_list_len = len(current_value_list)
    max_value = max(current_value_list)
    for current_digit in range(max_value, 0, -1):
      try:
        found_index = current_value_list.index(current_digit)
        remaining_digits = target_length - len(built_str)
        if (current_list_len - found_index + 1) > remaining_digits:
          built_str += str(current_digit)
          value_str = value_str[found_index + 1:]
          break
      except:
        if not quiet: print('ho ho ho')
        continue
    if len(built_str) == target_length:
      return int(built_str)

def find_batteries_deluxe(banks, num_batteries, quiet=True):
  jolts = [find_highest_jolt_deluxe(bank, num_batteries, quiet) for bank in banks]
  return sum(jolts)

find_batteries_deluxe(example, 12)
```
