## Example

```python
example = ['987654321111111','811111111111119','234234234234278','818181911112111']
banks = [list(b) for b in example]

def get_highest_jolt(bank):
  bank_size = len(bank)
  largest_jolt = 0
  for i in range(0, bank_size):
    for j in range(bank_size - 1, i, -1):
      current_jolt = int(bank[i] + bank[j])
      largest_jolt = current_jolt if current_jolt > largest_jolt else largest_jolt
  return largest_jolt

def find_batteries(banks):
  jolts = [get_highest_jolt(bank) for bank in banks]
  return sum(jolts)
    
find_batteries(banks)
```

```python

def find_greatest_number(numbers, number_length):
  len_numbers = len(numbers)
  num_str = ''
  allowed_skips = len_numbers - number_length
  current_number = int(numbers[0])
  for i in range(1, len_numbers + 1):
    next_number = int(numbers[i]) if i < len_numbers else 0
    if (current_number == next_number and i == 1): # edge case
      num_str += str(current_number)
    if (next_number >= current_number and allowed_skips > 0):
      allowed_skips -= 1
    else:
      num_str += str(current_number)
    current_number = next_number
  valid = len(num_str) == number_length
  return int(num_str), valid  

def get_highest_jolt_deluxe(bank, num_batteries):
  val, is_valid = find_greatest_number(bank, num_batteries)
  if (is_valid == False):
    val, is_valid = find_greatest_number(bank[::-1], num_batteries)
    val = int(str(val)[::-1])
  return val
  
 

def find_batteries_deluxe(banks, num_batteries):
  jolts = [get_highest_jolt_deluxe(bank, num_batteries) for bank in banks]
  return sum(jolts)

find_batteries_deluxe(example, 12)
```
